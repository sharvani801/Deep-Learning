## EXPERIMENT 6 : IMPLEMENTATION OF LOSS FUNCTIONS IN NEURAL NETWORKS

#Objective: ####
The objective of this experiment is to understand how neural networks measure prediction errors using different loss functions. It involves implementing Mean Squared Error (MSE) and Cross-Entropy Loss in Python, comparing their results, and understanding their applications in regression and classification problems.

Task 1: Define Actual Class Labels and Predicted Outputs

Create a simple classification dataset with actual class labels.

Define predicted probabilities for each class.

Display the actual and predicted values.

```python
import numpy as np

# Actual class labels
y_true = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_pred = np.array([0.9, 0.2, 0.8, 0.7, 0.1])

print("Actual Values:", y_true)
print("Predicted Values:", y_pred)
```

Task 2: Calculate Mean Squared Error and Cross-Entropy Loss

Write Python code to calculate Mean Squared Error (MSE).

Implement Cross-Entropy Loss using NumPy.

Calculate and display both loss values.

```python
import numpy as np

# Actual class labels
y_true = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_pred = np.array([0.9, 0.2, 0.8, 0.7, 0.1])

# Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)

# Cross-Entropy Loss
epsilon = 1e-15
y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_true * np.log(y_pred) +
    (1 - y_true) * np.log(1 - y_pred)
)

print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

Task 3: Compare the Results and Explain Their Usage

Compare the calculated MSE and Cross-Entropy Loss values.

Explain how MSE and Cross-Entropy measure prediction errors.

Identify when MSE and Cross-Entropy Loss are used.

Display the comparison results and summarize the differences.

```python
print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)

print("\nComparison:")
print("MSE measures the average squared difference between actual and predicted values.")
print("Cross-Entropy measures the difference between actual labels and predicted probabilities.")

print("\nUsage:")
print("MSE is commonly used for regression problems.")
print("Cross-Entropy is commonly used for classification problems.")
```

Result:

Mean Squared Error (MSE) and Cross-Entropy Loss were successfully implemented and calculated using Python. The experiment demonstrated how both loss functions measure prediction errors.

Conclusion:

The experiment helped us understand the differences between MSE and Cross-Entropy Loss. MSE is commonly used for regression problems, whereas Cross-Entropy Loss is widely used for classification problems in neural networks.
